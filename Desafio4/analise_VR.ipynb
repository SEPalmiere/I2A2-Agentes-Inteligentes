## Codigo Gerado pela IA com base na Análise Exploratória

```python
# =============================================================================
# SISTEMA DE AUTOMAÇÃO DE COMPRA DE VR/VA - VERSÃO FINAL CORRIGIDA
# =============================================================================

import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

print("🍽️ SISTEMA DE AUTOMAÇÃO DE VR/VA - VERSÃO CORRIGIDA")
print("=" * 80)

class VRAutomationSystem:
    """Sistema completo para automação do cálculo e compra de VR/VA"""
    
    def __init__(self):
        self.competencia = "05/2025"
        self.datasets = {}
        self.base_consolidada = None
        self.resultado_final = None
        
        self.config = {
            'percentual_empresa': 0.80,
            'percentual_funcionario': 0.20,
            'dia_corte_desligamento': 15,
            'cargos_excluidos': ['DIRETOR', 'ESTAGIARIO', 'APRENDIZ'],
            'situacoes_afastamento': ['LICENÇA MATERNIDADE', 'AFASTAMENTO', 'LICENÇA MÉDICA']
        }
    
    def carregar_datasets(self):
        """Carrega todos os datasets necessários"""
        print("\n📂 CARREGANDO DATASETS...")
        print("-" * 50)
        
        try:
            # 1. Base de colaboradores ativos
            self.datasets['ativos'] = pd.read_excel('ATIVOS.xlsx', sheet_name='ATIVOS')
            print(f"✅ ATIVOS: {len(self.datasets['ativos'])} registros")
            
            # 2. Férias
            self.datasets['ferias'] = pd.read_excel('FERIAS.xlsx', sheet_name='Planilha1')
            print(f"✅ FÉRIAS: {len(self.datasets['ferias'])} registros")
            
            # 3. Desligados
            self.datasets['desligados'] = pd.read_excel('DESLIGADOS.xlsx', sheet_name='DESLIGADOS ')
            print(f"✅ DESLIGADOS: {len(self.datasets['desligados'])} registros")
            
            # 4. Admissões do mês
            self.datasets['admissoes'] = pd.read_excel('ADMISSAO ABRIL.xlsx', sheet_name='Planilha1')
            print(f"✅ ADMISSÕES: {len(self.datasets['admissoes'])} registros")
            
            # 5. Base de valores por sindicato
            self.datasets['valores_sindicato'] = pd.read_excel('Base sindicato x valor.xlsx', sheet_name='Planilha1')
            print(f"✅ VALORES SINDICATO: {len(self.datasets['valores_sindicato'])} registros")
            
            # 6. Dias úteis - LEITURA ESPECIAL
            self.carregar_dias_uteis()
            
            # 7. Afastamentos
            self.datasets['afastamentos'] = pd.read_excel('AFASTAMENTOS.xlsx', sheet_name='Planilha1')
            print(f"✅ AFASTAMENTOS: {len(self.datasets['afastamentos'])} registros")
            
            # 8. Estagiários
            self.datasets['estagiarios'] = pd.read_excel('ESTAGIO.xlsx', sheet_name='Planilha1')
            print(f"✅ ESTAGIÁRIOS: {len(self.datasets['estagiarios'])} registros")
            
            # 9. Aprendizes
            self.datasets['aprendizes'] = pd.read_excel('APRENDIZ.xlsx', sheet_name='Planilha1')
            print(f"✅ APRENDIZES: {len(self.datasets['aprendizes'])} registros")
            
            # 10. Exterior
            self.datasets['exterior'] = pd.read_excel('EXTERIOR.xlsx', sheet_name='Planilha1')
            print(f"✅ EXTERIOR: {len(self.datasets['exterior'])} registros")
            
            return True
            
        except Exception as e:
            print(f"❌ Erro ao carregar datasets: {str(e)}")
            import traceback
            traceback.print_exc()
            return False
    
    def carregar_dias_uteis(self):
        """Carrega o arquivo de dias úteis com tratamento especial"""
        try:
            print("🔍 Carregando dias úteis com análise especial...")
            
            # Ler arquivo completo sem header
            df_raw = pd.read_excel('Base dias uteis.xlsx', sheet_name='Planilha1', header=None)
            print(f"   Dados brutos: {df_raw.shape}")
            print(f"   Conteúdo:\n{df_raw}")
            
            # Processar linha por linha para identificar dados válidos
            dados_validos = []
            
            for i in range(len(df_raw)):
                try:
                    linha = df_raw.iloc[i]
                    col1 = linha.iloc[0] if len(linha) > 0 else None
                    col2 = linha.iloc[1] if len(linha) > 1 else None
                    
                    # Pular se for header ou vazio
                    if (pd.isna(col1) or pd.isna(col2) or
                        str(col1).upper() in ['SINDICATO', 'SINDICADO', 'BASE'] or
                        str(col2).upper() in ['DIAS UTEIS', 'DIAS', 'VALOR']):
                        print(f"   ⏭️ Pulando linha {i}: '{col1}' | '{col2}' (header/inválido)")
                        continue
                    
                    # Tentar converter col2 para número
                    try:
                        dias_str = str(col2).replace(',', '.').strip()
                        if dias_str.replace('.', '').isdigit():
                            dias_num = float(dias_str)
                            dados_validos.append({
                                'SINDICATO': str(col1).strip(),
                                'DIAS_UTEIS': dias_num
                            })
                            print(f"   ✅ Linha {i}: '{col1}' -> {dias_num} dias")
                        else:
                            print(f"   ❌ Linha {i}: '{col2}' não é número válido")
                    except:
                        print(f"   ❌ Linha {i}: Erro ao converter '{col2}' para número")
                        
                except Exception as e:
                    print(f"   ❌ Erro na linha {i}: {e}")
                    continue
            
            # Criar DataFrame com dados válidos
            if dados_validos:
                self.datasets['dias_uteis'] = pd.DataFrame(dados_validos)
                print(f"✅ DIAS ÚTEIS: {len(self.datasets['dias_uteis'])} registros válidos")
                print(f"   Dados processados:\n{self.datasets['dias_uteis']}")
            else:
                # Criar dados padrão se não encontrar nada
                print("⚠️ Nenhum dado válido encontrado, criando padrão")
                self.datasets['dias_uteis'] = pd.DataFrame({
                    'SINDICATO': ['PADRÃO'],
                    'DIAS_UTEIS': [22]
                })
                print(f"✅ DIAS ÚTEIS: {len(self.datasets['dias_uteis'])} registros (padrão)")
                
        except Exception as e:
            print(f"❌ Erro crítico ao carregar dias úteis: {e}")
            # Criar dados de emergência
            self.datasets['dias_uteis'] = pd.DataFrame({
                'SINDICATO': ['EMERGENCIA'],
                'DIAS_UTEIS': [22]
            })
            print(f"✅ DIAS ÚTEIS: {len(self.datasets['dias_uteis'])} registros (emergência)")
    
    def limpar_e_padronizar_dados(self):
        """Limpa e padroniza os dados carregados"""
        print("\n🧹 LIMPEZA E PADRONIZAÇÃO DOS DADOS...")
        print("-" * 50)
        
        padronizacao_colunas = {
            'MATRICULA ': 'MATRICULA',
            'Cadastro': 'MATRICULA',
            'DATA DEMISSÃO': 'DATA_DEMISSAO',
            'COMUNICADO DE DESLIGAMENTO': 'COMUNICADO_DESLIGAMENTO',
            'TITULO DO CARGO': 'CARGO',
            'DESC. SITUACAO': 'SITUACAO',
            'DIAS DE FÉRIAS': 'DIAS_FERIAS',
            'Admissão': 'DATA_ADMISSAO'
        }
        
        for dataset_name, df in self.datasets.items():
            if df is not None and len(df) > 0:
                # Renomear colunas
                df.columns = [padronizacao_colunas.get(col, col) for col in df.columns]
                
                # Limpar espaços em branco
                for col in df.columns:
                    if df[col].dtype == 'object':
                        df[col] = df[col].astype(str).str.strip()
                
                # Converter MATRICULA para string padronizada
                if 'MATRICULA' in df.columns:
                    df['MATRICULA'] = df['MATRICULA'].astype(str).str.zfill(5)
                
                self.datasets[dataset_name] = df
        
        print("✅ Dados limpos e padronizados")
    
    def criar_base_consolidada(self):
        """Cria a base consolidada única"""
        print("\n🔗 CRIANDO BASE CONSOLIDADA...")
        print("-" * 50)
        
        base = self.datasets['ativos'].copy()
        print(f"📊 Base inicial (ATIVOS): {len(base)} registros")
        
        # Adicionar informações de data de admissão
        if 'DATA_ADMISSAO' not in base.columns:
            admissoes_mes = self.datasets['admissoes'][['MATRICULA', 'DATA_ADMISSAO']].copy()
            base = base.merge(admissoes_mes, on='MATRICULA', how='left')
        
        # Marcar colaboradores em férias
        ferias = self.datasets['ferias'][['MATRICULA', 'SITUACAO', 'DIAS_FERIAS']].copy()
        ferias['EM_FERIAS'] = True
        base = base.merge(ferias[['MATRICULA', 'EM_FERIAS', 'DIAS_FERIAS']], 
                         on='MATRICULA', how='left')
        base['EM_FERIAS'] = base['EM_FERIAS'].fillna(False)
        
        # Marcar desligados
        desligados = self.datasets['desligados'][['MATRICULA', 'DATA_DEMISSAO', 'COMUNICADO_DESLIGAMENTO']].copy()
        desligados['DESLIGADO'] = True
        base = base.merge(desligados, on='MATRICULA', how='left')
        base['DESLIGADO'] = base['DESLIGADO'].fillna(False)
        
        # Marcar afastamentos
        afastados = self.datasets['afastamentos'][['MATRICULA', 'SITUACAO']].copy()
        afastados = afastados[afastados['SITUACAO'].isin(self.config['situacoes_afastamento'])]
        afastados['AFASTADO'] = True
        afastados = afastados[['MATRICULA', 'AFASTADO']].drop_duplicates()
        base = base.merge(afastados, on='MATRICULA', how='left')
        base['AFASTADO'] = base['AFASTADO'].fillna(False)
        
        # Marcar estagiários e aprendizes
        estagiarios = self.datasets['estagiarios']['MATRICULA'].tolist()
        aprendizes = self.datasets['aprendizes']['MATRICULA'].tolist()
        base['ESTAGIARIO'] = base['MATRICULA'].isin(estagiarios)
        base['APRENDIZ'] = base['MATRICULA'].isin(aprendizes)
        
        # Marcar colaboradores no exterior
        exterior = self.datasets['exterior']['MATRICULA'].tolist()
        base['NO_EXTERIOR'] = base['MATRICULA'].isin(exterior)
        
        self.base_consolidada = base
        print(f"✅ Base consolidada criada: {len(base)} registros")
        
        return base
    
    def aplicar_exclusoes(self):
        """Remove colaboradores não elegíveis"""
        print("\n❌ APLICANDO EXCLUSÕES...")
        print("-" * 50)
        
        base = self.base_consolidada.copy()
        total_inicial = len(base)
        
        # Excluir por cargo
        mask_cargo = base['CARGO'].str.contains('|'.join(self.config['cargos_excluidos']), 
                                               case=False, na=False)
        print(f"🚫 Excluídos por cargo: {mask_cargo.sum()} registros")
        
        # Excluir estagiários e aprendizes
        print(f"🚫 Excluídos estagiários: {base['ESTAGIARIO'].sum()} registros")
        print(f"🚫 Excluídos aprendizes: {base['APRENDIZ'].sum()} registros")
        print(f"🚫 Excluídos afastados: {base['AFASTADO'].sum()} registros")
        print(f"🚫 Excluídos exterior: {base['NO_EXTERIOR'].sum()} registros")
        
        # Aplicar exclusões
        base_filtrada = base[
            ~mask_cargo & 
            (base['ESTAGIARIO'] == False) & 
            (base['APRENDIZ'] == False) & 
            (base['AFASTADO'] == False) & 
            (base['NO_EXTERIOR'] == False)
        ].copy()
        
        total_excluidos = total_inicial - len(base_filtrada)
        print(f"📊 Total excluído: {total_excluidos} registros")
        print(f"✅ Base elegível: {len(base_filtrada)} registros")
        
        self.base_consolidada = base_filtrada
        return base_filtrada
    
    def calcular_dias_uteis_por_colaborador(self):
        """Calcula os dias úteis VERSÃO CORRIGIDA"""
        print("\n📅 CALCULANDO DIAS ÚTEIS POR COLABORADOR...")
        print("-" * 50)
        
        # Criar mapeamento de dias úteis
        dias_uteis_map = {}
        df_dias = self.datasets['dias_uteis']
        
        print("🔍 Processando dados de dias úteis:")
        print(f"   DataFrame:\n{df_dias}")
        
        # Processar cada linha de forma segura
        for _, row in df_dias.iterrows():
            try:
                sindicato = str(row['SINDICATO']).strip()
                dias = float(row['DIAS_UTEIS'])
                dias_uteis_map[sindicato] = dias
                print(f"   ✅ {sindicato}: {dias} dias")
            except Exception as e:
                print(f"   ❌ Erro ao processar linha: {e}")
                continue
        
        # Valor padrão
        dias_uteis_padrao = 22
        
        if not dias_uteis_map:
            print("⚠️ Usando valor padrão para todos os sindicatos")
            dias_uteis_map = {'DEFAULT': dias_uteis_padrao}
        
        print(f"📊 Mapeamento final: {dias_uteis_map}")
        
        # Aplicar aos colaboradores
        base = self.base_consolidada.copy()
        
        # Função para mapear sindicato para dias úteis
        def obter_dias_uteis(sindicato):
            if pd.isna(sindicato):
                return dias_uteis_padrao
            
            sindicato_str = str(sindicato).strip()
            
            # Tentar match direto
            if sindicato_str in dias_uteis_map:
                return dias_uteis_map[sindicato_str]
            
            # Tentar match parcial
            for key, value in dias_uteis_map.items():
                if key.upper() in sindicato_str.upper() or sindicato_str.upper() in key.upper():
                    return value
            
            # Usar padrão
            return dias_uteis_padrao
        
        base['DIAS_UTEIS_SINDICATO'] = base['Sindicato'].apply(obter_dias_uteis)
        base['DIAS_UTEIS_FINAL'] = base['DIAS_UTEIS_SINDICATO']
        
        # Aplicar regras especiais
        # 1. Férias
        mask_ferias = base['EM_FERIAS'] == True
        if mask_ferias.any():
            base.loc[mask_ferias, 'DIAS_UTEIS_FINAL'] = (
                base.loc[mask_ferias, 'DIAS_UTEIS_SINDICATO'] - 
                base.loc[mask_ferias, 'DIAS_FERIAS'].fillna(0)
            ).clip(lower=0)
        
        # 2. Desligados
        mask_desligados = base['DESLIGADO'] == True
        if mask_desligados.any():
            base['DATA_DEMISSAO'] = pd.to_datetime(base['DATA_DEMISSAO'], errors='coerce')
            
            # Desligados até dia 15 com OK = 0 dias
            mask_deslig_ate_15_ok = (
                mask_desligados & 
                (base['DATA_DEMISSAO'].dt.day <= self.config['dia_corte_desligamento']) &
                (base['COMUNICADO_DESLIGAMENTO'].str.upper() == 'OK')
            )
            base.loc[mask_deslig_ate_15_ok, 'DIAS_UTEIS_FINAL'] = 0
            
            # Desligados após dia 15 = proporcional
            mask_deslig_apos_15 = (
                mask_desligados & 
                (base['DATA_DEMISSAO'].dt.day > self.config['dia_corte_desligamento'])
            )
            base.loc[mask_deslig_apos_15, 'DIAS_UTEIS_FINAL'] = (
                base.loc[mask_deslig_apos_15, 'DATA_DEMISSAO'].dt.day / 30 * 
                base.loc[mask_deslig_apos_15, 'DIAS_UTEIS_SINDICATO']
            )
        
        # 3. Admitidos no mês
        base['DATA_ADMISSAO'] = pd.to_datetime(base['DATA_ADMISSAO'], errors='coerce')
        mask_admitidos = base['DATA_ADMISSAO'].notna()
        if mask_admitidos.any():
            dias_restantes = 30 - base.loc[mask_admitidos, 'DATA_ADMISSAO'].dt.day + 1
            base.loc[mask_admitidos, 'DIAS_UTEIS_FINAL'] = (
                dias_restantes / 30 * base.loc[mask_admitidos, 'DIAS_UTEIS_SINDICATO']
            )
        
        # Garantir valores não negativos
        base['DIAS_UTEIS_FINAL'] = base['DIAS_UTEIS_FINAL'].clip(lower=0).round(2)
        
        self.base_consolidada = base
        
        print(f"✅ Dias úteis calculados para {len(base)} colaboradores")
        print(f"📊 Média de dias úteis: {base['DIAS_UTEIS_FINAL'].mean():.2f}")
        
        return base
    
    def calcular_valores_vr(self):
        """Calcula os valores de VR"""
        print("\n💰 CALCULANDO VALORES DE VR...")
        print("-" * 50)
        
        # Mapeamento de valores por estado
        valores_map = {}
        df_valores = self.datasets['valores_sindicato']
        
        print("🔍 Processando valores por estado:")
        print(f"   DataFrame:\n{df_valores.head()}")
        
        for _, row in df_valores.iterrows():
            try:
                estado = str(row.iloc[0]).strip()
                valor_raw = str(row.iloc[1])
                
                # Limpar valor
                valor_clean = valor_raw.replace('R$', '').replace(' ', '').replace(',', '.')
                valor = float(valor_clean)
                
                valores_map[estado] = valor
                print(f"   ✅ {estado}: R$ {valor:.2f}")
                
            except Exception as e:
                print(f"   ❌ Erro ao processar valor: {e}")
                continue
        
        valor_padrao = 35.00
        if not valores_map:
            valores_map = {'DEFAULT': valor_padrao}
        
        print(f"📊 Mapeamento de valores: {valores_map}")
        
        base = self.base_consolidada.copy()
        
        # Mapear estado
        def extrair_estado(sindicato):
            if pd.isna(sindicato):
                return list(valores_map.keys())[0]
            
            sindicato_upper = str(sindicato).upper()
            
            if 'SP' in sindicato_upper:
                return 'São Paulo'
            elif 'RJ' in sindicato_upper:
                return 'Rio de Janeiro'
            elif 'PR' in sindicato_upper:
                return 'Paraná'
            elif 'RS' in sindicato_upper:
                return 'Rio Grande do Sul'
            else:
                # Usar primeiro estado disponível
                return list(valores_map.keys())[0]
        
        base['ESTADO'] = base['Sindicato'].apply(extrair_estado)
        base['VALOR_DIARIO_VR'] = base['ESTADO'].map(valores_map).fillna(valor_padrao)
        
        # Calcular valores
        base['TOTAL_VR'] = (base['DIAS_UTEIS_FINAL'] * base['VALOR_DIARIO_VR']).round(2)
        base['CUSTO_EMPRESA'] = (base['TOTAL_VR'] * self.config['percentual_empresa']).round(2)
        base['DESCONTO_FUNCIONARIO'] = (base['TOTAL_VR'] * self.config['percentual_funcionario']).round(2)
        
        # Observações
        base['OBS_GERAL'] = ''
        base.loc[base['EM_FERIAS'] == True, 'OBS_GERAL'] = 'EM FÉRIAS'
        base.loc[base['DESLIGADO'] == True, 'OBS_GERAL'] = 'DESLIGADO'
        base.loc[base['DATA_ADMISSAO'].notna(), 'OBS_GERAL'] = 'ADMITIDO NO MÊS'
        
        self.base_consolidada = base
        
        print(f"✅ Valores calculados")
        print(f"💰 Valor total VR: R$ {base['TOTAL_VR'].sum():,.2f}")
        print(f"🏢 Custo empresa: R$ {base['CUSTO_EMPRESA'].sum():,.2f}")
        
        return base
    
    def gerar_planilha_final(self):
        """Gera a planilha final"""
        print("\n📊 GERANDO PLANILHA FINAL...")
        print("-" * 50)
        
        base = self.base_consolidada.copy()
        
        resultado = pd.DataFrame()
        resultado['Matricula'] = base['MATRICULA']
        resultado['Admissão'] = base['DATA_ADMISSAO'].dt.strftime('%m/%d/%y') if 'DATA_ADMISSAO' in base.columns else ''
        resultado['Sindicato do Colaborador'] = base['Sindicato']
        resultado['Competência'] = self.competencia
        resultado['Dias'] = base['DIAS_UTEIS_FINAL'].round(2)
        resultado['VALOR DIÁRIO VR'] = base['VALOR_DIARIO_VR'].round(2)
        resultado['TOTAL'] = base['TOTAL_VR'].round(2)
        resultado['Custo empresa'] = base['CUSTO_EMPRESA'].round(2)
        resultado['Desconto profissional'] = base['DESCONTO_FUNCIONARIO'].round(2)
        resultado['OBS GERAL'] = base['OBS_GERAL']
        
        # Filtrar apenas colaboradores com dias > 0
        resultado = resultado[resultado['Dias'] > 0].reset_index(drop=True)
        resultado = resultado.sort_values('Matricula').reset_index(drop=True)
        
        self.resultado_final = resultado
        
        print(f"✅ Planilha final gerada com {len(resultado)} colaboradores")
        print(f"💰 Valor total final: R$ {resultado['TOTAL'].sum():,.2f}")
        
        return resultado
    
    def salvar_resultado(self, nome_arquivo=None):
        """Salva o resultado"""
        if nome_arquivo is None:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M")
            nome_arquivo = f"VR_MENSAL_{self.competencia.replace('/', '')}_FINAL_{timestamp}.xlsx"
        
        try:
            with pd.ExcelWriter(nome_arquivo, engine='openpyxl') as writer:
                self.resultado_final.to_excel(writer, sheet_name='VR MENSAL', index=False)
                self.base_consolidada.to_excel(writer, sheet_name='Base Consolidada', index=False)
            
            print(f"✅ Arquivo salvo: {nome_arquivo}")
            return nome_arquivo
            
        except Exception as e:
            print(f"❌ Erro ao salvar: {e}")
            return None
    
    def executar_processo_completo(self):
        """Executa todo o processo"""
        print("🚀 INICIANDO PROCESSO COMPLETO")
        print("=" * 80)
        
        try:
            if not self.carregar_datasets():
                return False
            
            self.limpar_e_padronizar_dados()
            self.criar_base_consolidada()
            self.aplicar_exclusoes()
            self.calcular_dias_uteis_por_colaborador()
            self.calcular_valores_vr()
            self.gerar_planilha_final()
            
            arquivo_final = self.salvar_resultado()
            
            print("\n🎉 PROCESSO CONCLUÍDO COM SUCESSO!")
            print("=" * 80)
            print(f"📁 Arquivo: {arquivo_final}")
            print(f"👥 Colaboradores: {len(self.resultado_final)}")
            print(f"💰 Valor total: R$ {self.resultado_final['TOTAL'].sum():,.2f}")
            
            return True
            
        except Exception as e:
            print(f"\n❌ ERRO NO PROCESSO: {str(e)}")
            import traceback
            traceback.print_exc()
            return False

# =============================================================================
# FUNÇÕES DE DEBUG E VALIDAÇÃO
# =============================================================================

def debug_dias_uteis():
    """Debug específico do arquivo de dias úteis"""
    print("🔧 DEBUG - DIAS ÚTEIS")
    print("=" * 50)
    
    try:
        df = pd.read_excel('Base dias uteis.xlsx', sheet_name='Planilha1', header=None)
        print("📊 Conteúdo completo:")
        print(df)
        print()
        
        for i in range(len(df)):
            row = df.iloc[i]
            print(f"Linha {i}: {list(row)}")
            
    except Exception as e:
        print(f"❌ Erro: {e}")

def validar_arquivos():
    """Valida se todos os arquivos estão presentes"""
    arquivos = [
        'ATIVOS.xlsx', 'FERIAS.xlsx', 'DESLIGADOS.xlsx',
        'ADMISSAO ABRIL.xlsx', 'Base sindicato x valor.xlsx',
        'Base dias uteis.xlsx', 'AFASTAMENTOS.xlsx',
        'ESTAGIO.xlsx', 'APRENDIZ.xlsx', 'EXTERIOR.xlsx'
    ]
    
    print("🔍 VALIDANDO ARQUIVOS:")
    todos_ok = True
    
    for arquivo in arquivos:
        try:
            pd.read_excel(arquivo, nrows=1)
            print(f"✅ {arquivo}")
        except:
            print(f"❌ {arquivo}")
            todos_ok = False
    
    return todos_ok

# =============================================================================
# INSTRUÇÕES DE USO
# =============================================================================

print("""
🎯 COMO USAR:

1. VALIDAR ARQUIVOS:
   validar_arquivos()

2. DEBUG DIAS ÚTEIS:
   debug_dias_uteis()

3. EXECUTAR SISTEMA:
   sistema = VRAutomationSystem()
   sistema.executar_processo_completo()

4. VER RESULTADO:
   print(sistema.resultado_final.head())
""")

# EXECUTAR AGORA (descomente para usar):
# sistema = VRAutomationSystem()
# sistema.executar_processo_completo()
```
